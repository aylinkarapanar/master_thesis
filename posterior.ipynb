```python
import numpy as np
import pandas as pd
import pickle
from pathlib import Path
from scipy import linalg
from scipy.stats import multivariate_normal

PARAM_NAMES = {
    1: ["b0", "bint"],
    2: ["bext"],
    3: ["b0", "bint", "bext", "z"],
    4: ["b0", "bint", "bext"],
    5: ["guess"],
    6: ["bias1"],
    7: ["bias2"],
}


def check_and_invert(H):
    dets = np.array([linalg.det(H[i]) for i in range(H.shape[0])])
    bad  = np.where(np.abs(dets) < 1e-10)[0]
    if len(bad) > 0:
        print(f" Singular Hessian for subjects: {bad + 1}")
    else:
        print(f" All Hessians invertible")
    return np.array([linalg.inv(H[i]) for i in range(H.shape[0])])


def sample_posterior(P, Hinv, model_number, n_samples=1000, base_seed=1234):
    param_names = PARAM_NAMES[model_number]
    n_subjects  = P.shape[0]
    cols        = {}

    for i in range(n_subjects):
        samples = multivariate_normal.rvs(
            mean=P[i], cov=Hinv[i], size=n_samples, random_state=base_seed + i
        )
        samples = np.atleast_2d(samples)
        if samples.shape[0] != n_samples:
            samples = samples.T

        for j, pname in enumerate(param_names):
            cols[f"{pname}[{i + 1}]"] = samples[:, j]

    return pd.DataFrame(cols)


def generate_all_posteriors(hbi_output_dir="./hbi_output",
                            param_est_dir="./parameter_estimates/hbi",
                            n_samples=1000,
                            base_seed=1234):

    hbi_output_dir = Path(hbi_output_dir)
    param_est_dir  = Path(param_est_dir)

    # Iterate over seeds
    for seed_dir in sorted(hbi_output_dir.iterdir()):
        if not seed_dir.is_dir():
            continue
        seed_name = seed_dir.name
        print(f"\nSeed: {seed_name}")

        # Iterate over datasets within each seed
        for dataset_dir in sorted(seed_dir.iterdir()):
            if not dataset_dir.is_dir():
                continue
            base_name = dataset_dir.name  # e.g. 150_180_equal
            print(f"  Dataset: {base_name}")

            out_dir = param_est_dir / seed_name / "samples"
            out_dir.mkdir(parents=True, exist_ok=True)

            # Iterate over models
            for k, param_names in PARAM_NAMES.items():
                pkl_path = dataset_dir / f"cbm_m{k}.pkl"
                if not pkl_path.exists():
                    print(f"    Missing: {pkl_path}")
                    continue

                with open(pkl_path, "rb") as f:
                    cbm_k = pickle.load(f)

                P = np.array(cbm_k.output.parameters)       # (n_subjects, n_params)
                H = np.array(cbm_k.output.math.hessian)     # (n_subjects, n_params, n_params)

                print(f"    Model {k} ({', '.join(param_names)}): {P.shape[0]} subjects")

                Hinv = check_and_invert(H)
                df   = sample_posterior(P, Hinv, k, n_samples=n_samples, base_seed=base_seed)

                out_path = out_dir / f"{base_name}_posterior_{k}.csv"
                df.to_csv(out_path, index=False)
                print(f" Saved: {out_path}")


if __name__ == "__main__":
    generate_all_posteriors()
```
